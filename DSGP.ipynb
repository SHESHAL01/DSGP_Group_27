```python
import pandas as pd
import re
```

```python
filePath = "/content/drive/MyDrive/DSGP Group 27 /Dataset for RS/Udemy_Courses.csv"
df = pd.read_csv(filePath)

tech_skills = [
    'python', 'java', 'javascript', 'html', 'css', 'sql', 'mysql', 'mongodb',
    'react', 'angular', 'node', 'express', 'django', 'flask', 'pandas', 'numpy',
    'machine learning', 'deep learning', 'ai', 'artificial intelligence',
    'data science', 'data analysis', 'power bi', 'tableau', 'excel',
    'git', 'github', 'docker', 'kubernetes', 'aws', 'azure', 'cloud', 'linux',
    'tensorflow', 'pytorch', 'nlp', 'natural language processing', 'opencv',
    'cybersecurity', 'blockchain', 'devops', 'api', 'rest', 'spring', 'bootstrap',
    'php', 'laravel', 'c++', 'c#', 'r', 'scala', 'go', 'typescript', 'swift', 'kotlin'
]

def clean_text(text):
    if pd.isna(text):
        return ""
    text = text.lower()
    text = re.sub(r'[^a-z0-9\s+#]', ' ', text)
    return text

def extract_skills(Summary):
    found = []
    text = clean_text(Summary)
    for skill in tech_skills:
        pattern = r'\b' + re.escape(skill.lower()) + r'\b'
        if re.search(pattern, text):
            found.append(skill)
    return list(set(found))


#Apply the extraction
df["extracted_skills"] = df["Summary"].apply(extract_skills)

output_df = df[["Title", "Link", "extracted_skills"]]

output_df.to_csv("Udemy_Courses_with_skills.csv", index=False)

DS1 = output_df


```

```python
DS2 = pd.read_csv("/content/drive/MyDrive/DSGP Group 27 /Dataset for RS/coursera_course_2024.csv")
DS3 = pd.read_csv("/content/drive/MyDrive/DSGP Group 27 /Dataset for RS/coursera_course_dataset_v3.csv")
```

```python
DS1_clean = DS1.rename(columns={
    "Title": "Title",
    "extracted_skills": "Skills",
    "Link": "Url"
})[["Title", "Skills", "Url"]]

DS2_clean = DS2.rename(columns={
    "title": "Title",
    "Skills": "Skills",
    "URL": "Url"
})[["Title", "Skills", "Url"]]

DS3_clean = DS3.rename(columns={
    "Title": "Title",
    "Skills": "Skills",
    "course_url": "Url"
})[["Title", "Skills", "Url"]]

final_DS = pd.concat([DS1_clean, DS2_clean, DS3_clean], ignore_index=True)

```

```python
#week 9
final_DS = final_DS.dropna()

final_DS = final_DS.reset_index(drop=True)

print(f"Final dataset shape: {final_DS.shape}")
print(f"Null values check:\n{final_DS.isnull().sum()}")
```

```python

# Remove records with empty skills
final_DS = final_DS[
    (final_DS['Skills'] != "[]") &
    (final_DS['Skills'] != "['']") &
    (final_DS['Skills'].str.strip() != "") &
    (final_DS['Skills'].str.len() > 1)
]

final_DS = final_DS.drop_duplicates(subset=['Url'])

final_DS = final_DS.reset_index(drop=True)

print(f"Final dataset shape: {final_DS.shape}")
print(f"Sample of Skills column:\n{final_DS['Skills'].head()}")
```
